```python
import pandas as pd
import matplotlib.pyplot as plt
```

Läs in all data

```python
df = pd.read_csv("backend/data/solar.csv")
df.head()
```

Grundläggande info

```python
print("Antal rader och kolumner:", df.shape)
df.info()
```

Missing values

```python
df.isna().sum()
```

Sammanfattning

```python
df.describe()
```

Förmörkelsetyp

```python
df["Eclipse Type"].value_counts()
```

Antal solförmörkelser per typ

```python
fig, ax = plt.subplots(figsize=(8, 5))
df["Eclipse Type"].value_counts().plot(kind="bar", ax=ax, color="steelblue")
ax.set_title("Antal solförmörkelser per typ")
ax.set_xlabel("Typ")
ax.set_ylabel("Antal")
plt.tight_layout()
plt.show()
```

Fördelning av eclipse magnitude (hur stor del av solen som har täckts)

```python
fig, ax = plt.subplots(figsize=(8, 5))
df["Eclipse Magnitude"].plot(kind="hist", bins=30, ax=ax, color="steelblue", edgecolor="black")
ax.set_title("Fördelning av Eclipse Magnitude")
ax.set_xlabel("Eclipse Magnitude")
ax.set_ylabel("Antal förmörkelser")
plt.tight_layout()
plt.show()
```
